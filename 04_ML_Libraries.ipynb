```python
# 1. NumPy

NumPy stands for Numerical Python. It is a fundamental Python
library for numerical computing.

## Main Uses

- Numerical calculations
- One-dimensional and multidimensional arrays
- Matrix operations
- Statistical calculations
- Mathematical operations
- Linear algebra
- Random number generation

## How It Works

NumPy provides an efficient multidimensional array object called
`ndarray`. Operations can be performed on complete arrays without
writing explicit Python loops for every element.

## Important Functions

| Function | Purpose |
|---|---|
| `np.array()` | Creates an array |
| `np.zeros()` | Creates an array of zeros |
| `np.ones()` | Creates an array of ones |
| `np.arange()` | Creates a sequence of values |
| `np.mean()` | Calculates mean |
| `np.median()` | Calculates median |
| `np.std()` | Calculates standard deviation |
| `np.sum()` | Calculates sum |
| `np.min()` | Finds minimum |
| `np.max()` | Finds maximum |
```

```python
import numpy as np

# Create a NumPy array
data = np.array([10, 20, 30, 40, 50])

print("Array:", data)
print("Data type:", data.dtype)
print("Shape:", data.shape)

print("\nMean:", np.mean(data))
print("Median:", np.median(data))
print("Standard deviation:", np.std(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
print("Sum:", np.sum(data))

# Create other arrays
print("\nZeros:", np.zeros(5))
print("Ones:", np.ones(5))
print("Range:", np.arange(1, 6))
```

```python
# 2. Pandas

Pandas is a Python library used for data manipulation,
data cleaning, and data analysis.

## Main Uses

- Reading datasets
- Creating tables
- Data cleaning
- Handling missing values
- Filtering data
- Sorting data
- Grouping data
- Statistical summaries
- Working with CSV and other tabular files

## Main Data Structures

### Series

A one-dimensional labelled data structure.

### DataFrame

A two-dimensional table consisting of rows and columns.

## Important Functions

| Function | Purpose |
|---|---|
| `pd.DataFrame()` | Creates a DataFrame |
| `pd.read_csv()` | Reads CSV data |
| `df.head()` | Displays first rows |
| `df.tail()` | Displays last rows |
| `df.info()` | Displays dataset information |
| `df.describe()` | Generates statistical summary |
| `df.isnull()` | Detects missing values |
| `df.fillna()` | Fills missing values |
| `df.dropna()` | Removes missing values |
| `df.groupby()` | Groups data |
```

```python
import pandas as pd

# Create example data
data = {
    "Gene": ["GeneA", "GeneB", "GeneC", "GeneD", "GeneE"],
    "Expression": [10.5, 20.3, None, 15.8, 25.1],
    "Group": ["Control", "Treatment", "Control", "Treatment", "Treatment"]
}

df = pd.DataFrame(data)

print("Original Data:")
display(df)

print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

# Fill missing expression value with mean
df["Expression"] = df["Expression"].fillna(
    df["Expression"].mean()
)

print("\nData after handling missing values:")
display(df)

print("\nStatistical summary:")
display(df.describe())

print("\nMean expression by group:")
display(
    df.groupby("Group")["Expression"].mean()
)
```

```python
# 5. Scikit-learn

Scikit-learn is a widely used Python library for machine
learning.

## Main Uses

- Classification
- Regression
- Clustering
- Feature selection
- Data preprocessing
- Dimensionality reduction
- Model evaluation

## Classification Algorithms

- Logistic Regression
- Support Vector Machine
- K-Nearest Neighbors
- Decision Tree
- Random Forest
- Naive Bayes

## Clustering Algorithms

- K-Means
- DBSCAN
- Agglomerative Clustering

## Important Functions and Classes

| Function/Class | Purpose |
|---|---|
| `train_test_split()` | Splits data |
| `StandardScaler()` | Standardizes features |
| `LabelEncoder()` | Encodes labels |
| `LogisticRegression()` | Classification |
| `SVC()` | Support Vector Machine |
| `KNeighborsClassifier()` | KNN classification |
| `DecisionTreeClassifier()` | Decision tree |
| `RandomForestClassifier()` | Random forest |
| `KMeans()` | Clustering |
| `PCA()` | Dimensionality reduction |
| `accuracy_score()` | Calculates accuracy |
| `confusion_matrix()` | Evaluates classification |
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix

# Load Iris dataset
iris = load_iris()

X = iris.data
y = iris.target

print("Feature matrix shape:", X.shape)
print("Target shape:", y.shape)

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Standardize features
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Create model
model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Train model
model.fit(X_train_scaled, y_train)

# Predict
predictions = model.predict(X_test_scaled)

# Evaluate
accuracy = accuracy_score(
    y_test,
    predictions
)

cm = confusion_matrix(
    y_test,
    predictions
)

print("\nAccuracy:", round(accuracy, 4))

print("\nConfusion Matrix:")
print(cm)
```

```python
# 6. SciPy

SciPy stands for Scientific Python. It is used for scientific,
mathematical, and statistical computing.

SciPy extends NumPy by providing additional scientific
algorithms and functions.

## Main Uses

- Statistical analysis
- Hypothesis testing
- Optimization
- Probability distributions
- Signal processing
- Linear algebra
- Scientific calculations

## Important Functions

| Function | Purpose |
|---|---|
| `stats.ttest_ind()` | Independent t-test |
| `stats.ttest_rel()` | Paired t-test |
| `stats.pearsonr()` | Pearson correlation |
| `stats.spearmanr()` | Spearman correlation |
| `stats.mannwhitneyu()` | Mann-Whitney U test |
| `stats.chi2_contingency()` | Chi-square test |
```

```python
from scipy import stats

# Two example groups
group1 = [10, 12, 11, 13, 15]
group2 = [20, 21, 19, 22, 23]

# Independent t-test
t_statistic, p_value = stats.ttest_ind(
    group1,
    group2
)

print("Group 1:", group1)
print("Group 2:", group2)

print("\nT-statistic:", round(t_statistic, 4))
print("P-value:", round(p_value, 6))
```

```python
# 7. MLxtend

MLxtend stands for Machine Learning Extensions.

It provides additional machine learning utilities that
complement scikit-learn.

## Main Uses

- Association rule mining
- Frequent itemset mining
- Apriori algorithm
- Sequential feature selection
- Classification visualization
- Model evaluation

## Important Functions

| Function | Purpose |
|---|---|
| `apriori()` | Finds frequent itemsets |
| `association_rules()` | Generates association rules |
| `SequentialFeatureSelector()` | Performs feature selection |
| `plot_decision_regions()` | Visualizes decision regions |

## Apriori

The Apriori algorithm identifies item combinations that occur
frequently in a transaction dataset.

For example, it can identify products that are frequently
purchased together.
```

```python

import pandas as pd

from mlxtend.frequent_patterns import (
    apriori,
    association_rules
)

# Example transaction dataset
transactions = pd.DataFrame({
    "Milk":  [1, 1, 0, 1, 1, 0],
    "Bread": [1, 1, 1, 0, 1, 1],
    "Eggs":  [0, 1, 1, 1, 1, 0],
    "Butter":[0, 1, 0, 1, 1, 1]
}).astype(bool)

print("Transaction data:")
display(transactions)

# Find frequent itemsets
frequent_itemsets = apriori(
    transactions,
    min_support=0.5,
    use_colnames=True
)

print("Frequent itemsets:")
display(frequent_itemsets)

# Generate association rules
if len(frequent_itemsets) >= 2:

    rules = association_rules(
        frequent_itemsets,
        metric="confidence",
        min_threshold=0.6
    )

    print("Association rules:")

    if len(rules) > 0:
        display(
            rules[
                [
                    "antecedents",
                    "consequents",
                    "support",
                    "confidence",
                    "lift"
                ]
            ]
        )
    else:
        print("No association rules met the selected threshold.")

else:
    print("Not enough frequent itemsets to generate rules.")
```

```python
# 8. MiniSom

MiniSom is a lightweight Python implementation of a
Self-Organizing Map (SOM).

A Self-Organizing Map is an unsupervised neural network that
maps high-dimensional data onto a lower-dimensional grid.

## Main Uses

- Clustering
- Pattern recognition
- Data visualization
- Dimensionality reduction
- Exploring similarities between samples

## How It Works

A SOM consists of neurons arranged in a grid.

During training:

1. An input sample is presented to the network.
2. The Best Matching Unit (BMU) is identified.
3. The BMU and neighboring neurons are updated.
4. Repeated training causes similar samples to be mapped
   to nearby neurons.

## Important Functions

| Function | Purpose |
|---|---|
| `MiniSom()` | Creates SOM |
| `random_weights_init()` | Initializes weights |
| `train_random()` | Trains SOM using random samples |
| `train_batch()` | Batch training |
| `winner()` | Finds Best Matching Unit |
| `distance_map()` | Calculates neuron distances |
```

```python
# If MiniSom is not installed, run:
# !pip install minisom

import numpy as np
import matplotlib.pyplot as plt
from minisom import MiniSom

# Example dataset
data = np.array([
    [0.1, 0.2],
    [0.2, 0.1],
    [0.15, 0.25],
    [0.8, 0.9],
    [0.9, 0.8],
    [0.85, 0.95]
])

# Create SOM
som = MiniSom(
    x=3,
    y=3,
    input_len=2,
    sigma=1.0,
    learning_rate=0.5,
    random_seed=42
)

# Initialize weights
som.random_weights_init(data)

# Train SOM
som.train_random(
    data,
    num_iteration=100
)

# Display winning neurons
print("Winning neurons:")

for sample in data:
    print(
        sample,
        "->",
        som.winner(sample)
    )

# Display distance map
plt.figure(figsize=(6, 5))

plt.imshow(
    som.distance_map().T,
    cmap="bone_r"
)

plt.colorbar(label="Distance")

plt.title("MiniSom Self-Organizing Map")
plt.xlabel("SOM X coordinate")
plt.ylabel("SOM Y coordinate")

plt.tight_layout()
plt.show()
```
